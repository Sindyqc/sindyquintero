```python
Estudiantes_notas = [
    ("Camila", 20, 4.5),
    ("Jose", 18, 3.8),
    ("Andres", 22, 4.2),
    ("Maria", 25, 5.0),
    ("Ricardo", 18, 4.1),
    ("Sofia", 19, 5.0),
    ("Mariana", 19, 4.5),
    ("Ana", 20, 3.9)
]

# Para cada estudiante de la lista, mostrar su edad y su nota
for estudiante in Estudiantes_notas:
    print(f"{estudiante[0]} tiene {estudiante[1]} años y obtuvo una nota de {estudiante[2]}")
```

```python
for estudiante in Estudiantes_notas:
    nota = estudiante[2]

    if nota >= 4.5:
        clasificacion = "Excelente"
    elif nota >=4.0:
        clasificacion = "Bueno"
    elif nota >= 3.0:
        clasificacion = "Aceptable"
    else: 
        clasificacion = "Reprobo"

    print (f"{estudiante[0]} - {clasificacion}")
```

```python
suma_notas = 0
for estudiante in Estudiantes_notas:
    suma_notas += estudiante[2]

promedio = suma_notas / len(Estudiantes_notas)

print (f"El promedio general es: {promedio:.2f}")
```

```python
nombre_buscar =input("Ingrese el nombre del estudiante: ")

encontrado = False

for estudiante in Estudiantes_notas:
    if estudiante[0].lower() == nombre_buscar.lower():
        encontrado = True
        print(f"El estudiante {estudiante[0]} fue encontrado.")
        break

if not encontrado:
    print(f"No se encontro ningun estudiante con el nombre {nombre_buscar}.") 
```

```python
ciudades_estudiantes = {
    "Camila": "Barranquilla",
    "Jose": "Cartagena",
    "Andres": "Bogota",
    "Maria": "Medellin",
    "Ricardo": "Bogota",
    "Sofia": "Pereira",
    "Mariana": "Barranquilla",
    "Ana": "Bogota"
}

for estudiante, ciudad in ciudades_estudiantes.items():
    print(f"{estudiante} vive en {ciudad}")
```

```python
cantidad_por_ciudad ={}

for ciudad in ciudades_estudiantes.values():
    if ciudad in cantidad_por_ciudad:
        cantidad_por_ciudad[ciudad] += 1
    else:
        cantidad_por_ciudad[ciudad] = 1

print(cantidad_por_ciudad)
```

```python
cantidad = 0
suma = 0

numero = float(input("Ingrese otro numero (0 para terminar): "))

while numero != 0:
    cantidad += 1 
    suma += numero

    numero = float(input("Ingrese otro numero(0) para terminar): "))

    if cantidad > 0:
        promedio = suma / cantidad
    else:
        promedio = 0

    print(f"cantidad de numeros ingresados: {cantidad}")
    print(f"Suma total: {suma}")
    print(f"promedio de los numeros ingresados: {promedio:.2f}")
```
